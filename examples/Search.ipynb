# Wikidata

```python
from kif_lib import *
from kif_lib.vocabulary import wd
```

Create a Wikidata Search API

```python
kb = Search('wikidata') 
results = kb.item_descriptor('Brazil', limit=5)
display(*results)
```

Create a Search that uses the Wikidata Embeddings Project (Vector Database).

```python
kb = Search('wikidata-vdb', rerank=True) 
results = kb.item_descriptor('Brazil', limit=5)
display(*results)
```

```python
kb = Search('pubchem') 
results = kb.item_descriptor('benzene', limit=1)
display(*results)
```

```python
kb = Search('mixer', 'wikidata', 'wikidata-vdb', 'pubchem')
results = kb.item_descriptor('benzene', limit=5)
display(*results)
```
